```python
from pyspark.sql import functions as F
df_silver_lido = spark.read.table("silver_cotacoes")
```

```python
df_gold = df_silver_lido.groupby("code").agg(
    F.round(F.avg("bid"), 4).alias("avg_bid"), 
    F.max("high").alias("max_high"), 
    F.min("low").alias("min_low"), 
    F.count("code").alias("event_count")
    )
```

```python
df_gold.write.format("Delta").mode("overwrite").saveAsTable("gold_resumo_cambio")
```
