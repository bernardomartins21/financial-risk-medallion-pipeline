```python
import requests
import json

dbutils.widgets.text("api_token", "", "Token da API")
token = dbutils.widgets.get("api_token")

dbutils.widgets.text("pares_moedas", "", "Moedas")
pares_moedas = dbutils.widgets.get("pares_moedas")

url = f'https://economia.awesomeapi.com.br/json/last/{pares_moedas}?token={token}'

response = requests.get(url)

print(response.status_code)

dados_json = response.json()
```

```python
import pandas as pd

lista_dados = list(dados_json.values())

pdf_bronze = pd.DataFrame(lista_dados)

df_bronze = spark.createDataFrame(pdf_bronze)

#display(df_bronze)
```
